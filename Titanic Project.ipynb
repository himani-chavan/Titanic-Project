```python
import pandas as pd
```

```python
#importing titanic database
dataf = pd.read_csv('titanic_dataset.csv')
dataf
```

```python
#finding mean
dataf['Age'].mean()
```

```python
dataf['Age'].median()
```

```python
dataf['Age'].mode()
```

```python
#finding outliers
dataf['Age'].plot(kind='box')
```

```python
#fill out the missing values in age
#we will take median here because there are too many outliers present0
dataf['Age'] = dataf['Age'].fillna(dataf['Age'].median())
dataf
```

```python
#to check whether all the null values are filled
dataf['Age'].isnull().sum()
```

```python
dataf.isnull().sum()
```

```python
dataf['Embarked'].mode()
```

```python
dataf['Embarked']= dataf['Embarked'].fillna(dataf['Embarked'].mode()[0])
dataf
dataf['Embarked'].isnull().sum()
```
